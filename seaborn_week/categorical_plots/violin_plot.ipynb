# Basic Visualization of fmri dataset through violin plot

```python
import seaborn 
```

```python
seaborn.set(style='whitegrid')
```

```python
fmri = seaborn.load_dataset('fmri')
```

```python
seaborn.violinplot(x="timepoint",
                   y="signal",
                   data=fmri)
```

```python
tips = seaborn.load_dataset("tips")
```

```python
seaborn.violinplot(x=tips["total_bill"])
```

```python
seaborn.violinplot(x="tip", y="day", data=tips)
```

```python
seaborn.violinplot(x="day",y="total_bill", hue="time", data=tips)
```

```python
seaborn.violinplot(x='day', y='tip', data=tips, linewidth=4)
```

```python
seaborn.violinplot(x="day", y="total_bill", hue="smoker", 
                   data=tips, palette="Set2", dodge=True)
```

```python
seaborn.violinplot(x="time", y="tip", data=tips,
                   order=["Dinner", "Lunch"])
```
