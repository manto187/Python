# Basic Visualization of fmri dataset through violin plot

```python
import seaborn 
```

```python
seaborn.set(style='whitegrid')
```

```python
fmri = seaborn.load_dataset('fmri')
```

```python
seaborn.violinplot(x="timepoint",
                   y="signal",
                   data=fmri)
```

```python
tips = seaborn.load_dataset("tips")
```

```python
seaborn.violinplot(x=tips["total_bill"])
```
